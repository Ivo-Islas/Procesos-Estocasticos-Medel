# Caminata Aleatoria

Código que implementa una caminata aleatoria 1D, con probabilidad p de moverse a la derecha y probabilidad q de moverse a la izquierda, tal que $p + q = 1$.

```python
import random as r # agregamos la libreria que genera numeros aleatorios
import matplotlib.pyplot as plt # agregamos la libreria que nos permite graficar
```

```python
def caminata(p, n):
    x = 0                                           #   posicion inicial en 0
    h = [0]                                        # inicial la lista en el primer movimiento
    for i in range(n):                        # para cada uno de los pasos
        if r.random() < p:                   # si el numero es menor a la probabilidad dada
            x=x+1                                # se mueve 1 a la derecha
            print('avanza       (+1): ',x) #busco hacer el movimiento visible
        else:
            x=x-1                                # se mueve 1 a la izquierda
            print('retrocede  (-1): ',x) #busco hacer el movimiento visible
        h.append(x)                           # agrego la posicion a la lista h

    return h                                      # entrego la lista h
```

```python
pp = float(input("Introduce la probabilidad p (entre 0 y 1): "))  #pide probabilidad elegida por el lector
np = 100                                                                                    # el numero de pasos lo define en 100
cam = caminata(pp, np)                                                              # guarda el movimiento de la caminata en una variable
plt.plot(cam)                                                                             # grafica lo guardado en la variable definida en la linea anterior
plt.yticks(range(-5, 6, 2))                                                          # define los espacios principales mostrados y el paso entre estos
plt.axhline(0, color='red', linestyle=':', alpha=0.5)                   # grafica una linea horizontal para denotar la posicion inicial a lo largo del experimento
plt.show()                                                                                 # imprime la grafica simulada
```

```python
def dibujar_diagrama(matriz):
  G =nx.DiGraph ()
  letras = ['a', 'b', 'c','d','e','f','g']
  for i in range (matriz.rows):
    for j in range (matriz.cols):
      if matriz[i,j] != 0:
        G.add_edge(letras[i], letras[j], weight=matriz[i,j])
  pos = nx.spring_layout(G)
  nx.draw_networkx_nodes(G, pos)
  nx.draw_networkx_edges(G, pos)
  nx.draw_networkx_labels(G, pos)
  edge_labels = nx.get_edge_attributes(G, 'weight')
  nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)
  plt.show()
  return G
```
